## Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

## Random Sampling

Random sampling ensures that each observation has an equal chance of being selected.

This helps reduce bias and ensures representativeness.

```python
data = np.random.randn(1000)

sample = np.random.choice(data, size=100, replace=False)

print('Sample mean:', np.mean(sample))
```

## Sampling Bias

Bias occurs when the sample does not represent the population.

Example: selecting only a specific group may lead to misleading conclusions.

## Sampling Distribution

A sampling distribution is the distribution of a statistic (like mean) across many samples.

```python
means = []
for _ in range(1000):
    sample = np.random.choice(data, size=50)
    means.append(np.mean(sample))

plt.hist(means, bins=30)
plt.title('Sampling Distribution of the Mean')
plt.show()
```

## Central Limit Theorem (CLT)

The Central Limit Theorem states that the sampling distribution of the mean approaches a normal distribution as sample size increases, regardless of the original distribution.

```python
sample_sizes = [5, 20, 50]

for size in sample_sizes:
    means = [np.mean(np.random.choice(data, size=size)) for _ in range(1000)]
    plt.hist(means, bins=30, alpha=0.5, label=f'n={size}')

plt.legend()
plt.title('CLT Demonstration')
plt.show()
```

## Standard Error

Standard Error measures the variability of the sample mean.

Formula: std / sqrt(n)

```python
std_error = np.std(data) / np.sqrt(len(data))
print('Standard Error:', std_error)
```

## Bootstrap Method

Bootstrap involves resampling with replacement to estimate statistics.

```python
boot_means = []
for _ in range(1000):
    resample = np.random.choice(data, size=len(data), replace=True)
    boot_means.append(np.mean(resample))

plt.hist(boot_means, bins=30)
plt.title('Bootstrap Distribution')
plt.show()
```

## Confidence Interval

Confidence intervals estimate the range where a population parameter lies.

```python
ci = np.percentile(boot_means, [2.5, 97.5])
print('95% Confidence Interval:', ci)
```

## Normal Distribution

The normal distribution is a symmetric, bell-shaped distribution.

```python
x = np.linspace(-3, 3, 100)
y = stats.norm.pdf(x)

plt.plot(x, y)
plt.title('Normal Distribution')
plt.show()
```

## Binomial Distribution

Used for binary outcomes (success/failure).

```python
n, p = 10, 0.5
x = np.arange(0, n+1)
y = stats.binom.pmf(x, n, p)

plt.bar(x, y)
plt.title('Binomial Distribution')
plt.show()
```

## Poisson Distribution

Models count data over time or space.

```python
mu = 3
x = np.arange(0, 10)
y = stats.poisson.pmf(x, mu)

plt.bar(x, y)
plt.title('Poisson Distribution')
plt.show()
```
